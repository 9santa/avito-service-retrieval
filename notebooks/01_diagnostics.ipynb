```python
import re

def normalize_query(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()
    text = text.replace("ё", "е")
    text = re.sub(r"[^0-9a-zа-я]+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text


train["search_query_norm"] = (
    train["search_query"].map(normalize_query)
)

queries["search_query_norm"] = (
    queries["search_query"].map(normalize_query)
)

train_norm_queries = set(train["search_query_norm"])

seen_norm = queries[
    "search_query_norm"
].isin(train_norm_queries)

print("Seen normalized:", seen_norm.sum())
print("Seen normalized fraction:", seen_norm.mean())
```

### Насколько запрос определяет category / microcat

```python
def topk_coverage(df: pd.DataFrame, group_col, target_col, k):
    """
    Для каждой группы находит k самых частых target-значений
    и считает долю train-строк, попавших в эти top-k.
    """
    counts = (
        df.groupby([group_col, target_col], dropna=False)
        .size()
        .rename("count")
        .reset_index()
    )

    counts["rank"] = (
        counts.groupby(group_col)["count"]
        .rank(method="first", ascending=False)
    )

    return (
        counts.loc[counts["rank"] <= k, "count"].sum()
        / counts["count"].sum()
    )


for k in [1, 2, 3, 5]:
    coverage = topk_coverage(
        train,
        group_col="search_category",
        target_col="item_category_id",
        k=k,
    )

    print(f"search_category -> item_category top-{k}: {coverage:.4f}")
```

```python
for k in [1, 3, 5, 10, 20]:
    coverage = topk_coverage(
        train,
        group_col="search_category",
        target_col="item_microcat_id",
        k=k,
    )

    print(f"search_category -> microcat top-{k}: {coverage:.4f}")
```

```python
for k in [1, 3, 5, 10]:
    coverage = topk_coverage(
        train,
        group_col="search_query_norm",
        target_col="item_microcat_id",
        k=k,
    )

    print(f"query -> microcat top-{k}: {coverage:.4f}")
```

```python
query_counts = train["search_query_norm"].value_counts()

repeated_queries = query_counts[
    query_counts >= 5
].index

train_repeated = train[
    train["search_query_norm"].isin(repeated_queries)
].copy()

print("Unique repeated queries:", len(repeated_queries))
print("Rows:", len(train_repeated))
```

```python
query_microcat = (
    train_repeated
    .groupby(
        ["search_query_norm", "item_microcat_id"]
    )
    .size()
    .rename("count")
    .reset_index()
)

query_totals = (
    query_microcat
    .groupby("search_query_norm")["count"]
    .sum()
    .rename("total")
)

query_max = (
    query_microcat
    .groupby("search_query_norm")["count"]
    .max()
    .rename("top1")
)

query_purity = (
    pd.concat(
        [query_totals, query_max],
        axis=1,
    )
)

query_purity["top1_share"] = (
    query_purity["top1"]
    / query_purity["total"]
)

display(
    query_purity["top1_share"].describe(
        percentiles=[0.1,0.25,0.5,0.75,0.9, 0.95]
    )
)


for threshold in [0.5, 0.7, 0.8, 0.9, 1.0]:
    print(
        threshold,
        (query_purity["top1_share"] >= threshold).mean(),
    )
```
